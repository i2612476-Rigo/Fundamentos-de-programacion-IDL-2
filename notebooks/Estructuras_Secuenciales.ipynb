## **# EJERCICIO N° 1 — VELOCIDAD FINAL DE UN AUTO**
Enunciado: Determina la velocidad final de un auto si su velocidad inicial es 0 km/h y su aceleración es 0.8 m/s²; y el tiempo transcurrido es de 30 segundos.


```python
# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
velocidad_inicial = 0        # float | velocidad inicial en m/s
aceleracion       = 0.8      # float | aceleración en m/s²
tiempo            = 30       # float | tiempo transcurrido en segundos

# --- Cuerpo del algoritmo ---
# Fórmula cinemática: vf = v0 + a * t
velocidad_final     = velocidad_inicial + aceleracion * tiempo  # float | m/s
velocidad_final_kmh = velocidad_final * 3.6                     # float | km/h

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — VELOCIDAD FINAL DEL AUTO")
print("=" * 50)
print(f"  Velocidad inicial : {velocidad_inicial:.2f} m/s")
print(f"  Aceleración       : {aceleracion:.2f} m/s²")
print(f"  Tiempo            : {tiempo:.2f} s")
print(f"  Velocidad final   : {velocidad_final:.2f} m/s")
print(f"  Velocidad final   : {velocidad_final_kmh:.2f} km/h")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 2 — PROMEDIO DE CUATRO EXÁMENES**
Enunciado: Un estudiante realiza cuatro exámenes, los cuales tienen la misma ponderación. Representen el algoritmo correspondiente para obtener el promedio de las calificaciones obtenidas.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 2 — PROMEDIO DE CUATRO EXÁMENES       ║
# ║  Curso   : Fundamentos de Programación IDL2          ║
# ║  Tema    : Estructuras Secuenciales                  ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
nota1 = 15.0   # float | calificación del examen 1
nota2 = 18.0   # float | calificación del examen 2
nota3 = 12.0   # float | calificación del examen 3
nota4 = 17.0   # float | calificación del examen 4

# --- Cuerpo del algoritmo ---
def calcular_promedio(n1, n2, n3, n4):
    """Calcula el promedio de 4 calificaciones con igual ponderación."""
    return (n1 + n2 + n3 + n4) / 4

promedio = calcular_promedio(nota1, nota2, nota3, nota4)  # float | promedio final

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — PROMEDIO DE EXÁMENES")
print("=" * 50)
print(f"  Examen 1  : {nota1:.2f}")
print(f"  Examen 2  : {nota2:.2f}")
print(f"  Examen 3  : {nota3:.2f}")
print(f"  Examen 4  : {nota4:.2f}")
print(f"  Promedio  : {promedio:.2f}")
```

## **EJERCICIO N° 3 — PORCENTAJE DE INVERSIÓN DE TRES SOCIOS**
Enunciado: Tres personas deciden invertir su dinero para fundar una empresa. Cada una de ellas invierte una cantidad distinta. Obtener el porcentaje que cada uno invierte con respecto a la cantidad total invertida.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 3 — PORCENTAJE DE INVERSIÓN           ║
# ║  Curso   : Fundamentos de Programación IDL2          ║
# ║  Tema    : Estructuras Secuenciales                  ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
inversion_socio1 = 5000.0   # float | inversión del socio 1 en soles
inversion_socio2 = 3000.0   # float | inversión del socio 2 en soles
inversion_socio3 = 2000.0   # float | inversión del socio 3 en soles

# --- Cuerpo del algoritmo ---
def calcular_porcentaje(inversion, total):
    """Calcula el porcentaje de inversión respecto al total."""
    return (inversion / total) * 100

total_invertido = inversion_socio1 + inversion_socio2 + inversion_socio3  # float | soles
porcentaje1     = calcular_porcentaje(inversion_socio1, total_invertido)  # float | %
porcentaje2     = calcular_porcentaje(inversion_socio2, total_invertido)  # float | %
porcentaje3     = calcular_porcentaje(inversion_socio3, total_invertido)  # float | %

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — PORCENTAJE DE INVERSIÓN")
print("=" * 50)
print(f"  Total invertido   : S/ {total_invertido:.2f}")
print(f"  Socio 1 — S/ {inversion_socio1:.2f}  →  {porcentaje1:.2f}%")
print(f"  Socio 2 — S/ {inversion_socio2:.2f}  →  {porcentaje2:.2f}%")
print(f"  Socio 3 — S/ {inversion_socio3:.2f}  →  {porcentaje3:.2f}%")
print("=" * 50)
```

## **EJERCICIO N° 4 — SUELDO BASE + COMISIONES DEL VENDEDOR**
Enunciado: Un vendedor recibe su sueldo base más 25% extra por comisión de sus ventas; el vendedor desea saber cuánto dinero recibirá por concepto de comisiones por las tres ventas que realiza en el mes y el total que recibirá en el mes tomando en cuenta sueldo base y comisiones.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 4 — SUELDO BASE + COMISIONES          ║
# ║  Curso   : Fundamentos de Programación IDL1          ║
# ║  Tema    : Estructuras Secuenciales                  ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
sueldo_base   = 1500.0   # float | sueldo base mensual en soles
venta1        = 2000.0   # float | monto de la venta 1 en soles
venta2        = 3500.0   # float | monto de la venta 2 en soles
venta3        = 1800.0   # float | monto de la venta 3 en soles
tasa_comision = 0.25     # float | tasa de comisión (25%)

# --- Cuerpo del algoritmo ---
def calcular_comision(venta, tasa):
    """Calcula la comisión de una venta aplicando la tasa dada."""
    return venta * tasa

comision1      = calcular_comision(venta1, tasa_comision)   # float | comisión venta 1
comision2      = calcular_comision(venta2, tasa_comision)   # float | comisión venta 2
comision3      = calcular_comision(venta3, tasa_comision)   # float | comisión venta 3
total_comision = comision1 + comision2 + comision3          # float | total comisiones
sueldo_total   = sueldo_base + total_comision               # float | sueldo total del mes

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — LIQUIDACIÓN DEL VENDEDOR")
print("=" * 50)
print(f"  Sueldo base        : S/ {sueldo_base:.2f}")
print(f"  Comisión venta 1   : S/ {comision1:.2f}")
print(f"  Comisión venta 2   : S/ {comision2:.2f}")
print(f"  Comisión venta 3   : S/ {comision3:.2f}")
print(f"  Total comisiones   : S/ {total_comision:.2f}")
print(f"  SUELDO TOTAL MES   : S/ {sueldo_total:.2f}")
print("=" * 50)
```

## **EJERCICIO N° 5 — INTERÉS BANCARIO AL 2% MENSUAL**
Enunciado: Suponga que un individuo desea invertir su capital en un banco y desea saber cuánto dinero ganará después de un año si el banco paga a razón de 2% mensual.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 5 — INTERÉS BANCARIO AL 2% MENSUAL   ║
# ║  Curso   : Fundamentos de Programación IDL1         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
capital      = 10000.0  # float | capital invertido en soles
tasa_mensual = 0.02     # float | tasa de interés mensual (2%)
meses        = 12       # int   | número de meses en un año

# --- Cuerpo del algoritmo ---
def calcular_interes_simple(capital, tasa, meses):
    """Calcula el interés simple acumulado durante el período."""
    return capital * tasa * meses

interes_ganado = calcular_interes_simple(capital, tasa_mensual, meses)  # float | interés
monto_final    = capital + interes_ganado  # float | capital + interés al año

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — INVERSIÓN BANCARIA")
print("=" * 50)
print(f"  Capital inicial    : S/ {capital:.2f}")
print(f"  Tasa mensual       : {tasa_mensual * 100:.0f}%")
print(f"  Período            : {meses} meses")
print(f"  Interés ganado     : S/ {interes_ganado:.2f}")
print(f"  Monto final        : S/ {monto_final:.2f}")
print("=" * 50)
```

## **EJERCICIO N° 6 — EDAD EN MESES, SEMANAS, DÍAS Y HORAS**
Enunciado: Realice un programa que representen el algoritmo donde se ingrese la edad de una persona para determinar aproximadamente cuántos meses, semanas, días y horas ha vivido una persona.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 6 — EDAD EN MESES, SEMANAS, DÍAS Y HORAS ║
# ║  Curso   : Fundamentos de Programación IDL1         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
edad_anos = 25  # int | edad de la persona en años

# --- Cuerpo del algoritmo ---
def convertir_edad(anos):
    """Convierte la edad en años a meses, semanas, días y horas."""
    meses   = anos * 12
    semanas = anos * 52
    dias    = anos * 365
    horas   = dias * 24
    return meses, semanas, dias, horas

edad_meses, edad_semanas, edad_dias, edad_horas = convertir_edad(edad_anos)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — EDAD EN DISTINTAS UNIDADES")
print("=" * 50)
print(f"  Edad en años    : {edad_anos} años")
print(f"  Edad en meses   : {edad_meses} meses")
print(f"  Edad en semanas : {edad_semanas} semanas")
print(f"  Edad en días    : {edad_dias} días")
print(f"  Edad en horas   : {edad_horas} horas")
print("=" * 50)
```

## **EJERCICIO N° 7 — COBRO DE ESTACIONAMIENTO**
Enunciado: Un estacionamiento requiere determinar el cobro que debe aplicar a las personas que lo utilizan. Considere que el cobro es con base en las horas que lo disponen y que las fracciones de hora se toman como completas y realice un programa que permita determinar el cobro. El pago por hora es de S/5.00 y el pago por los minutos restantes de S/0.10 por minuto.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 7 — COBRO DE ESTACIONAMIENTO         ║
# ║  Curso   : Fundamentos de Programación IDL1         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
tiempo_minutos  = 95    # int   | tiempo total de uso en minutos
tarifa_hora     = 5.00  # float | tarifa por hora completa en soles
tarifa_minuto   = 0.10  # float | tarifa por minuto fraccionario en soles

# --- Cuerpo del algoritmo ---
def calcular_cobro(minutos, t_hora, t_minuto):
    """
    Calcula el cobro del estacionamiento.
    Horas completas a t_hora y minutos restantes a t_minuto.
    """
    horas_completas   = minutos // 60
    minutos_restantes = minutos % 60
    cobro_horas       = horas_completas * t_hora
    cobro_minutos     = minutos_restantes * t_minuto
    total             = cobro_horas + cobro_minutos
    return horas_completas, minutos_restantes, cobro_horas, cobro_minutos, total

horas, minutos_rest, pago_horas, pago_minutos, total_pagar = calcular_cobro(
    tiempo_minutos, tarifa_hora, tarifa_minuto
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COBRO DE ESTACIONAMIENTO")
print("=" * 50)
print(f"  Tiempo total      : {tiempo_minutos} minutos")
print(f"  Horas completas   : {horas} hora(s)  →  S/ {pago_horas:.2f}")
print(f"  Minutos restantes : {minutos_rest} min      →  S/ {pago_minutos:.2f}")
print(f"  TOTAL A PAGAR     : S/ {total_pagar:.2f}")
print("=" * 50)
```

## **EJERCICIO N° 8 — ÁREA DE TERRENO FORMA A (TRAPECIO)**
Enunciado: Una empresa constructora vende terrenos con la forma A de la figura. Obtener el área respectiva de un terreno de medidas ingresando el valor de los 3 lados.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 8 — ÁREA DE TERRENO FORMA A          ║
# ║  Curso   : Fundamentos de Programación IDL1         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
lado_a = 10.0   # float | base mayor del trapecio (metros)
lado_b = 15.0   # float | altura del trapecio (metros)
lado_c = 6.0    # float | base menor del trapecio (metros)

# --- Cuerpo del algoritmo ---
def calcular_area_trapecio(a, b, c):
    """
    Calcula el área de la Forma A (trapecio rectángulo).
    Fórmula: Área = ((A + C) / 2) * B
    A y C son las bases paralelas, B es la altura.
    """
    return ((a + c) / 2) * b

area_terreno = calcular_area_trapecio(lado_a, lado_b, lado_c)  # float | m²

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — ÁREA DEL TERRENO FORMA A")
print("=" * 50)
print(f"  Lado A (base mayor)  : {lado_a:.2f} m")
print(f"  Lado B (altura)      : {lado_b:.2f} m")
print(f"  Lado C (base menor)  : {lado_c:.2f} m")
print(f"  Área del terreno     : {area_terreno:.2f} m²")
print("=" * 50)
```

## **EJERCICIO N° 9 — COMPRA CON DESCUENTO Y VUELTO**
Enunciado: Cuanto pagaría un cliente por una compra con descuento y cuanto sería el cambio que recibiría en un pago en efectivo. Ingresar el monto base, el porcentaje de descuento y el con cuánto dinero está pagando el cliente.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 9 — COMPRA CON DESCUENTO Y VUELTO   ║
# ║  Curso   : Fundamentos de Programación IDL2         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
monto_base           = 250.0  # float | precio original en soles
porcentaje_descuento = 15.0   # float | porcentaje de descuento
pago_cliente         = 300.0  # float | dinero entregado por el cliente

# --- Cuerpo del algoritmo ---
def calcular_compra(monto, porcentaje, pago):
    """Calcula descuento, precio final y vuelto al cliente."""
    descuento    = monto * (porcentaje / 100)
    precio_final = monto - descuento
    vuelto       = pago - precio_final
    return descuento, precio_final, vuelto

descuento, precio_final, vuelto = calcular_compra(
    monto_base, porcentaje_descuento, pago_cliente
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COMPRA CON DESCUENTO")
print("=" * 50)
print(f"  Precio original    : S/ {monto_base:.2f}")
print(f"  Descuento ({porcentaje_descuento:.0f}%)    : S/ {descuento:.2f}")
print(f"  Precio final       : S/ {precio_final:.2f}")
print(f"  Pago del cliente   : S/ {pago_cliente:.2f}")
print(f"  Vuelto             : S/ {vuelto:.2f}")
print("=" * 50)
```

## **EJERCICIO N° 10 — CAJERO AUTOMÁTICO CON DENOMINACIONES ÓPTIMAS**
Enunciado: Se necesita saber la cantidad de dinero que un cajero automático debe proporcionar en diferentes denominaciones, teniendo en cuenta que el usuario debe ingresar la cantidad a retirar y el cajero tiene disponible las siguientes denominaciones de dinero: 200, 100, 50, 20, 10, 5, 1. Se debe buscar entregar la menor cantidad de dinero por parte del cajero.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 10 — CAJERO AUTOMÁTICO               ║
# ║  Curso   : Fundamentos de Programación IDL1         ║
# ║  Tema    : Estructuras Secuenciales                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
monto_retirar  = 385                      # int  | monto a retirar en soles
denominaciones = [200, 100, 50, 20, 10, 5, 1]  # list | billetes disponibles

# --- Cuerpo del algoritmo ---
def calcular_denominaciones(monto, billetes):
    """
    Determina la distribución óptima (mínima cantidad de billetes)
    usando el algoritmo greedy: mayor denominación primero.
    """
    distribucion = {}
    restante = monto
    for billete in billetes:
        cantidad = restante // billete
        if cantidad > 0:
            distribucion[billete] = cantidad
            restante = restante % billete
    return distribucion

distribucion = calcular_denominaciones(monto_retirar, denominaciones)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — CAJERO AUTOMÁTICO")
print("=" * 50)
print(f"  Monto a retirar   : S/ {monto_retirar}")
print("-" * 50)
print("  Denominación  |  Cantidad  |  Subtotal")
print("-" * 50)
total_billetes = 0
for billete, cantidad in distribucion.items():
    subtotal = billete * cantidad
    total_billetes += cantidad
    print(f"  S/ {billete:<10}  |  {cantidad:<10}  |  S/ {subtotal:.2f}")
print("-" * 50)
print(f"  Total billetes    : {total_billetes}")
print(f"  Monto entregado   : S/ {monto_retirar}")
print("=" * 50)

# ═══════════ FIN ═══════════
```
