## **EJERCICIO N° 1 — SUMA Y PROMEDIO DE PARES E IMPARES DEL 1 AL N**
Tipo: Estructura Repetitiva: for

Enunciado: Calcular la suma y promedio de los números pares e impares por separado, de 1 hasta N.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 1 — SUMA Y PROMEDIO PARES E IMPARES   ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
N              = 20      # int   | límite superior del rango
suma_pares     = 0       # int   | acumulador suma de pares
suma_impares   = 0       # int   | acumulador suma de impares
conteo_pares   = 0       # int   | contador de números pares
conteo_impares = 0       # int   | contador de números impares

# --- Cuerpo del algoritmo ---
# Función que calcula suma y promedio de pares e impares
def calcular_pares_impares(n):
    """
    Recorre del 1 a N y acumula suma y conteo
    de números pares e impares por separado.
    """
    s_par = 0; s_imp = 0
    c_par = 0; c_imp = 0
    for i in range(1, n + 1):          # Bucle for: recorre del 1 al N
        if i % 2 == 0:                 # Condición: número par
            s_par += i
            c_par += 1
        else:                          # Condición: número impar
            s_imp += i
            c_imp += 1
    prom_par = s_par / c_par if c_par > 0 else 0
    prom_imp = s_imp / c_imp if c_imp > 0 else 0
    return s_par, c_par, prom_par, s_imp, c_imp, prom_imp

suma_p, cont_p, prom_p, suma_i, cont_i, prom_i = calcular_pares_impares(N)

# --- Mostrar resultados ---
print("=" * 50)
print(f"   RESULTADO — PARES E IMPARES DEL 1 AL {N}")
print("=" * 50)
print(f"  PARES:")
print(f"    Cantidad : {cont_p}")
print(f"    Suma     : {suma_p}")
print(f"    Promedio : {prom_p:.2f}")
print(f"  IMPARES:")
print(f"    Cantidad : {cont_i}")
print(f"    Suma     : {suma_i}")
print(f"    Promedio : {prom_i:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 2 — CLASIFICAR 10 NÚMEROS**
Tipo: Estructura Repetitiva: for

Enunciado: Realizar un algoritmo que permita pedir 10 números y determine e imprima cuantos son pares, impares, positivos, neutros y negativos.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 2 — CLASIFICAR 10 NÚMEROS             ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
numeros     = [4, -3, 0, 7, -1, 2, 8, 0, -5, 6]  # list | 10 números ingresados
pares       = 0   # int | contador de pares
impares     = 0   # int | contador de impares
positivos   = 0   # int | contador de positivos
neutros     = 0   # int | contador de neutros (cero)
negativos   = 0   # int | contador de negativos

# --- Cuerpo del algoritmo ---
def clasificar_numeros(lista):
    """
    Recorre una lista de 10 números y clasifica cada uno
    como par/impar y como positivo/neutro/negativo.
    """
    c_par=0; c_imp=0; c_pos=0; c_neu=0; c_neg=0
    for num in lista:                  # Bucle for: recorre lista
        # Clasificar par o impar
        if num % 2 == 0:
            c_par += 1
        else:
            c_imp += 1
        # Clasificar positivo, neutro o negativo
        if num > 0:
            c_pos += 1
        elif num == 0:
            c_neu += 1
        else:
            c_neg += 1
    return c_par, c_imp, c_pos, c_neu, c_neg

pares, impares, positivos, neutros, negativos = clasificar_numeros(numeros)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — CLASIFICACIÓN DE 10 NÚMEROS")
print("=" * 50)
print(f"  Números ingresados : {numeros}")
print(f"  Pares              : {pares}")
print(f"  Impares            : {impares}")
print(f"  Positivos          : {positivos}")
print(f"  Neutros (cero)     : {neutros}")
print(f"  Negativos          : {negativos}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 3 — PORCENTAJES RESPECTO A UN NÚMERO X**
Tipo: Estructura Repetitiva: for

Enunciado: Hacer un programa que permita ingresar un conjunto de N números y un numero más X mostrar como resultado: Porcentaje de números menores que X, Porcentaje de números mayores de X, Porcentaje de números iguales de X, Porcentaje de números comprendidos entre 1 y X, Porcentaje de números que no cumple ninguna condición.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 3 — PORCENTAJES RESPECTO A X          ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
numeros    = [3, 7, 10, 2, 10, 15, 5, 8, 1, 10]  # list | conjunto de N números
X          = 10     # int | número de referencia para comparar
N          = len(numeros)  # int | cantidad total de números

# --- Cuerpo del algoritmo ---
def calcular_porcentajes(lista, x):
    """
    Calcula porcentajes de números menores, mayores,
    iguales, entre 1 y X, y los que no cumplen nada.
    """
    menores = 0; mayores = 0; iguales = 0
    entre   = 0; ninguna = 0
    n = len(lista)

    for num in lista:                  # Bucle for: recorre N números
        if num < x:
            menores += 1
        elif num > x:
            mayores += 1
        elif num == x:
            iguales += 1

        if 1 <= num <= x:
            entre += 1

        if num >= x or num <= 0:
            ninguna += 1

    pct_men = (menores / n) * 100
    pct_may = (mayores / n) * 100
    pct_igu = (iguales / n) * 100
    pct_ent = (entre   / n) * 100
    pct_nin = (ninguna / n) * 100
    return pct_men, pct_may, pct_igu, pct_ent, pct_nin

pct_m, pct_ma, pct_i, pct_e, pct_n = calcular_porcentajes(numeros, X)

# --- Mostrar resultados ---
print("=" * 50)
print(f"   RESULTADO — PORCENTAJES RESPECTO A X={X}")
print("=" * 50)
print(f"  Números            : {numeros}")
print(f"  Total números      : {N}")
print(f"  Menores que {X}    : {pct_m:.2f}%")
print(f"  Mayores que {X}    : {pct_ma:.2f}%")
print(f"  Iguales a {X}      : {pct_i:.2f}%")
print(f"  Entre 1 y {X}      : {pct_e:.2f}%")
print(f"  No cumple ninguna  : {pct_n:.2f}%")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 4 — MÁXIMO Y MÍNIMO DE N NÚMEROS**
Tipo: Estructura Repetitiva: for

Enunciado: De una lista de N números determinar simultáneamente el máximo y mínimo número.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 4 — MÁXIMO Y MÍNIMO DE N NÚMEROS     ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
numeros = [34, 7, 89, 12, 56, 3, 45, 78, 23, 67]  # list | lista de N números
maximo  = None   # int | valor máximo encontrado
minimo  = None   # int | valor mínimo encontrado

# --- Cuerpo del algoritmo ---
def encontrar_max_min(lista):
    """
    Recorre la lista simultáneamente buscando
    el valor máximo y mínimo sin usar max() ni min().
    """
    maximo = lista[0]   # Inicializar con primer elemento
    minimo = lista[0]
    for num in lista:   # Bucle for: recorre todos los números
        if num > maximo:
            maximo = num
        if num < minimo:
            minimo = num
    return maximo, minimo

maximo, minimo = encontrar_max_min(numeros)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — MÁXIMO Y MÍNIMO")
print("=" * 50)
print(f"  Lista      : {numeros}")
print(f"  Total      : {len(numeros)} números")
print(f"  Máximo     : {maximo}")
print(f"  Mínimo     : {minimo}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 5 — AHORRO ANUAL CON DEPÓSITOS MENSUALES**
Tipo: Estructura Repetitiva: for

Enunciado: Realizar un algoritmo para determinar cuánto ahorrará una persona en un año, si al final de cada mes deposita cantidades variables de dinero; se quiere saber cuánto lleva ahorrado cada vez que haga un deposito en cada mes y luego mostrar el ahorro final de 12 meses.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 5 — AHORRO ANUAL MENSUAL              ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
meses    = ["Enero","Febrero","Marzo","Abril","Mayo","Junio",
            "Julio","Agosto","Septiembre","Octubre","Noviembre","Diciembre"]
depositos = [200,350,150,400,300,250,500,180,420,310,270,380]  # list | depósito por mes
ahorro_acumulado = 0.0   # float | ahorro acumulado progresivo

# --- Cuerpo del algoritmo ---
def calcular_ahorro_anual(meses_lista, depositos_lista):
    """
    Recorre 12 meses mostrando el ahorro acumulado
    después de cada depósito mensual variable.
    """
    acumulado = 0.0
    print("-" * 45)
    print(f"  {'MES':<12} {'DEPÓSITO':>10} {'ACUMULADO':>12}")
    print("-" * 45)
    for i in range(12):                # Bucle for: 12 meses
        acumulado += depositos_lista[i]
        print(f"  {meses_lista[i]:<12} S/{depositos_lista[i]:>8.2f} S/{acumulado:>10.2f}")
    return acumulado

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — AHORRO ANUAL")
print("=" * 50)
ahorro_final = calcular_ahorro_anual(meses, depositos)
print("-" * 45)
print(f"  AHORRO FINAL AÑO  : S/ {ahorro_final:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 6 — CRONÓMETRO (HH:MM:SS)**
Tipo: Estructura Repetitiva: while / for anidado

Enunciado: Hacer un programa que muestre un cronometro (hh:mm:ss), indicando las horas, minutos y segundos (Considerar que el tiempo no será exacto al de un cronometro real solo será referencial).


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 6 — CRONÓMETRO (HH:MM:SS)            ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for anidado                           ║
# ╚══════════════════════════════════════════════════════╝

import time  # Módulo estándar para pausas

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
horas_max   = 0    # int | horas máximas del cronómetro (0=solo minutos)
minutos_max = 1    # int | minutos máximos a mostrar
segundos_max= 10   # int | segundos por minuto

# --- Cuerpo del algoritmo ---
def mostrar_cronometro(h_max, m_max, s_max):
    """
    Muestra un cronómetro referencial en formato HH:MM:SS
    usando bucles for anidados para horas, minutos y segundos.
    """
    print("  Cronómetro iniciado (referencial):")
    print("-" * 30)
    for h in range(h_max + 1):         # Bucle for: horas
        for m in range(m_max + 1):     # Bucle for anidado: minutos
            for s in range(s_max):     # Bucle for anidado: segundos
                print(f"  {h:02d}:{m:02d}:{s:02d}", end="\r")
                time.sleep(0.05)       # Pausa referencial (no exacta)
    print(f"  {h_max:02d}:{m_max:02d}:{s_max-1:02d} — Cronómetro finalizado")

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — CRONÓMETRO HH:MM:SS")
print("=" * 50)
mostrar_cronometro(horas_max, minutos_max, segundos_max)
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 7 — FACTORIAL Y PUNTO DE IGUALDAD**
Tipo: Estructura Repetitiva: for

Enunciado: Crea una aplicación que pida dos números y calcule su factorial de cada número, pero el programa también tiene que indicar hasta que valor serian iguales. Ejemplo: 7! > 1x2x3x4x5x6x7 / 10! > 1x2x3x4x5x6x7x8x9x10 / >> son iguales hasta el múltiplo 7


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 7 — FACTORIAL Y PUNTO DE IGUALDAD    ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle for                                   ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
num1 = 7    # int | primer número para calcular factorial
num2 = 10   # int | segundo número para calcular factorial

# --- Cuerpo del algoritmo ---
def calcular_factorial_detalle(n):
    """
    Calcula el factorial de n mostrando la serie paso a paso.
    Retorna el factorial y la lista de multiplicadores.
    """
    factorial  = 1
    serie      = []
    for i in range(1, n + 1):     # Bucle for: del 1 al n
        factorial *= i
        serie.append(i)
    return factorial, serie

def encontrar_igualdad(s1, s2):
    """
    Compara dos series y encuentra hasta qué valor
    comparten los mismos multiplicadores.
    """
    menor_serie = s1 if len(s1) <= len(s2) else s2
    ultimo_igual = 0
    for val in menor_serie:       # Bucle for: compara elemento a elemento
        if val in s2:
            ultimo_igual = val
    return ultimo_igual

fact1, serie1 = calcular_factorial_detalle(num1)
fact2, serie2 = calcular_factorial_detalle(num2)
punto_igualdad = encontrar_igualdad(serie1, serie2)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — FACTORIAL Y PUNTO DE IGUALDAD")
print("=" * 50)
serie1_str = "x".join(str(x) for x in serie1)
serie2_str = "x".join(str(x) for x in serie2)
print(f"  {num1}! = {serie1_str} = {fact1}")
print(f"  {num2}! = {serie2_str} = {fact2}")
print(f"  Son iguales hasta el múltiplo: {punto_igualdad}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 8 — ESTADÍSTICAS DE ASISTENTES A UNA FIESTA**
Tipo: Estructura Repetitiva: while

Enunciado: A una fiesta asistieron personas de diferentes edades y sexos. Construir un algoritmo donde se ingrese los datos de los N asistentes. Calcular: Cuantas personas asistieron a la fiesta, Cuantos hombres y cuantas mujeres, Promedio de edades por sexo, La edad de la persona de menor y mayor edad que asistió. Nota: El sistema debe permitir al menos el ingreso los datos de una persona y solicitar si se desea registrar a otra.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 8 — ESTADÍSTICAS DE FIESTA            ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas                     ║
# ║  Tipo  : Bucle while (al menos 1 registro)           ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
# Datos de asistentes (simulando ingreso del usuario)
asistentes = [
    {"nombre": "Ana",    "sexo": "F", "edad": 25},
    {"nombre": "Luis",   "sexo": "M", "edad": 32},
    {"nombre": "María",  "sexo": "F", "edad": 19},
    {"nombre": "Carlos", "sexo": "M", "edad": 45},
    {"nombre": "Rosa",   "sexo": "F", "edad": 28},
]

# --- Cuerpo del algoritmo ---
def calcular_estadisticas_fiesta(lista):
    """
    Procesa la lista de asistentes y calcula:
    total, hombres, mujeres, promedios de edad por sexo,
    menor y mayor edad asistida.
    """
    total     = len(lista)     # int   | total asistentes
    hombres   = 0              # int   | contador hombres
    mujeres   = 0              # int   | contador mujeres
    suma_h    = 0              # int   | suma edades hombres
    suma_m    = 0              # int   | suma edades mujeres
    menor_edad = lista[0]["edad"]   # int | menor edad
    mayor_edad = lista[0]["edad"]   # int | mayor edad

    i = 0
    while i < len(lista):      # Bucle while: recorre todos los asistentes
        p = lista[i]
        if p["sexo"] == "M":   # Condición: hombre
            hombres += 1
            suma_h  += p["edad"]
        else:                  # Condición: mujer
            mujeres += 1
            suma_m  += p["edad"]
        if p["edad"] < menor_edad:
            menor_edad = p["edad"]
        if p["edad"] > mayor_edad:
            mayor_edad = p["edad"]
        i += 1

    prom_h = suma_h / hombres if hombres > 0 else 0
    prom_m = suma_m / mujeres if mujeres > 0 else 0
    return total, hombres, mujeres, prom_h, prom_m, menor_edad, mayor_edad

total, hombres, mujeres, prom_h, prom_m, menor, mayor = calcular_estadisticas_fiesta(asistentes)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — ESTADÍSTICAS DE LA FIESTA")
print("=" * 50)
print(f"  Total asistentes       : {total}")
print(f"  Hombres                : {hombres}")
print(f"  Mujeres                : {mujeres}")
print(f"  Promedio edad hombres  : {prom_h:.2f} años")
print(f"  Promedio edad mujeres  : {prom_m:.2f} años")
print(f"  Edad menor asistente   : {menor} años")
print(f"  Edad mayor asistente   : {mayor} años")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 9 — PRODUCTO CON ARRAY: 100x98x96x...x2**
Tipo: Estructura Repetitiva: for con array

Enunciado: Calcular el producto utilizando arrays: 100 x 98 x 96 x 94 x . . . x 1 en este orden.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 9 — PRODUCTO CON ARRAY                ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas + Array             ║
# ║  Tipo  : Bucle for con lista (array)                 ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
# Array: números pares de 100 hasta 2 (descendente)
array_pares  = list(range(100, 0, -2))  # list | [100,98,96,...,2]
producto     = 1                         # int  | acumulador del producto

# --- Cuerpo del algoritmo ---
def calcular_producto_array(array):
    """
    Calcula el producto de todos los elementos del array
    usando un bucle for. Muestra los primeros valores.
    """
    prod = 1
    for num in array:              # Bucle for: recorre el array
        prod *= num
    return prod

producto = calcular_producto_array(array_pares)

# Mostrar primeros y últimos elementos del array
primeros = array_pares[:5]
ultimos  = array_pares[-5:]

# --- Mostrar resultados ---
print("=" * 55)
print("   RESULTADO — PRODUCTO CON ARRAY")
print("=" * 55)
print(f"  Array     : {primeros} ... {ultimos}")
print(f"  Total elementos : {len(array_pares)}")
print(f"  Operación : 100 x 98 x 96 x ... x 2")
print(f"  Producto  : {producto}")
print("=" * 55)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 10 — ESTADÍSTICAS DE NOTAS DE N ESTUDIANTES**

Enunciado: Desarrolla un programa que para N (con un mínimo de 5 y máximo de 20) estudiantes permita leer sus notas (calificación) obtenidas en un curso. Se debe validar que la nota tenga un valor entre 0 y 20. Se sabe que la mínima nota aprobatoria es 13. Al final calcule: a) La mayor nota, b) La menor nota, c) La cantidad de alumnos que obtuvieron la mayor nota, d) La cantidad de alumnos que obtuvieron la menor nota, e) El promedio de notas del salón, f) La cantidad de notas pares, g) La cantidad de notas impares, h) La cantidad de alumnos aprobados, i) La cantidad de alumnos desaprobados, j) El porcentaje de alumnos aprobados, k) El porcentaje de alumnos desaprobados.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 10 — ESTADÍSTICAS DE NOTAS            ║
# ║  Curso : Fundamentos de Programación IDL3            ║
# ║  Tema  : Estructuras Repetitivas + Array             ║
# ║  Tipo  : Bucle for con validación while              ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
N      = 8       # int  | número de estudiantes (mín 5, máx 20)
notas  = [15, 12, 18, 9, 13, 17, 11, 13]  # list | notas validadas 0-20
APROBATORIO = 13  # int  | nota mínima aprobatoria

# --- Cuerpo del algoritmo ---
def calcular_estadisticas_notas(lista_notas, aprobatorio):
    """
    Calcula todas las estadísticas requeridas para N estudiantes:
    mayor, menor, cant_mayor, cant_menor, promedio,
    pares, impares, aprobados, desaprobados y porcentajes.
    """
    n          = len(lista_notas)
    mayor_nota = lista_notas[0]    # int   | mayor nota
    menor_nota = lista_notas[0]    # int   | menor nota
    suma       = 0                 # int   | suma total notas
    cant_mayor = 0                 # int   | alumnos con mayor nota
    cant_menor = 0                 # int   | alumnos con menor nota
    pares      = 0                 # int   | notas pares
    impares    = 0                 # int   | notas impares
    aprobados  = 0                 # int   | alumnos aprobados
    desaprobados = 0               # int   | alumnos desaprobados

    # Primer recorrido: encontrar mayor y menor
    for nota in lista_notas:       # Bucle for: buscar max y min
        if nota > mayor_nota:
            mayor_nota = nota
        if nota < menor_nota:
            menor_nota = nota

    # Segundo recorrido: calcular todas las estadísticas
    for nota in lista_notas:       # Bucle for: calcular estadísticas
        suma += nota
        if nota == mayor_nota:
            cant_mayor += 1
        if nota == menor_nota:
            cant_menor += 1
        if nota % 2 == 0:
            pares += 1
        else:
            impares += 1
        if nota >= aprobatorio:
            aprobados += 1
        else:
            desaprobados += 1

    promedio    = suma / n
    pct_aprobados    = (aprobados    / n) * 100
    pct_desaprobados = (desaprobados / n) * 100

    return (mayor_nota, menor_nota, cant_mayor, cant_menor,
            promedio, pares, impares, aprobados, desaprobados,
            pct_aprobados, pct_desaprobados)

(mayor, menor, c_may, c_men, prom, pares, impares,
 aprobados, desaprobados, pct_apr, pct_des) = calcular_estadisticas_notas(notas, APROBATORIO)

# --- Mostrar resultados ---
print("=" * 55)
print("   RESULTADO — ESTADÍSTICAS DE NOTAS")
print("=" * 55)
print(f"  Estudiantes          : {N}")
print(f"  Notas                : {notas}")
print(f"  a) Mayor nota        : {mayor}")
print(f"  b) Menor nota        : {menor}")
print(f"  c) Alumnos c/mayor   : {c_may}")
print(f"  d) Alumnos c/menor   : {c_men}")
print(f"  e) Promedio          : {prom:.2f}")
print(f"  f) Notas pares       : {pares}")
print(f"  g) Notas impares     : {impares}")
print(f"  h) Aprobados         : {aprobados}")
print(f"  i) Desaprobados      : {desaprobados}")
print(f"  j) % Aprobados       : {pct_apr:.2f}%")
print(f"  k) % Desaprobados    : {pct_des:.2f}%")
print("=" * 55)

# ═══════════ FIN ═══════════
```
