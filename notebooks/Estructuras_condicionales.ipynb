## **EJERCICIO N° 1 — COBRO EN LLANTERA**
Tipo de condición: Condición Compuesta (if / else)

Enunciado: Calcular el total que una persona debe pagar en una llantera, si el precio de cada llanta es de S/150 si se compran menos de 5 llantas y de S/120 si se compran 5 o más.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 1 — COBRO EN LLANTERA                 ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Condición Compuesta (if / else)             ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
cantidad_llantas = 6       # int   | número de llantas a comprar
precio_normal    = 150.0   # float | precio por llanta (menos de 5)
precio_descuento = 120.0   # float | precio por llanta (5 o más)

# --- Cuerpo del algoritmo ---
# Condición Compuesta: if / else
def calcular_pago_llantera(cantidad, p_normal, p_descuento):
    """Calcula el total a pagar según cantidad de llantas."""
    if cantidad < 5:                          # Condición Simple
        precio_unitario = p_normal
        tipo_precio     = "precio normal"
    else:                                     # Condición Compuesta
        precio_unitario = p_descuento
        tipo_precio     = "precio con descuento"
    total = cantidad * precio_unitario
    return precio_unitario, tipo_precio, total

precio_unit, tipo_precio, total_pagar = calcular_pago_llantera(
    cantidad_llantas, precio_normal, precio_descuento
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COBRO EN LLANTERA")
print("=" * 50)
print(f"  Cantidad de llantas : {cantidad_llantas}")
print(f"  Tipo de precio      : {tipo_precio}")
print(f"  Precio unitario     : S/ {precio_unit:.2f}")
print(f"  TOTAL A PAGAR       : S/ {total_pagar:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 2 — MENOR Y MAYOR DE TRES NÚMEROS**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: ¿Cómo se podría determinar cuál de los 3 números brindados por usuario es el menor y mayor de todos ellos?


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 2 — MENOR Y MAYOR DE TRES NÚMEROS    ║
# ║  Curso : Fundamentos de Programación IDL2           ║
# ║  Tema  : Estructuras Condicionales                  ║
# ║  Tipo  : Selección Múltiple (if / elif / else)      ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
num1 = 45    # int | primer número ingresado por usuario
num2 = 12    # int | segundo número ingresado por usuario
num3 = 78    # int | tercer número ingresado por usuario

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else
def encontrar_menor_mayor(a, b, c):
    """Determina el menor y mayor de tres números."""
    # Determinar el menor
    if a <= b and a <= c:       # Condición Simple
        menor = a
    elif b <= a and b <= c:     # Selección Múltiple
        menor = b
    else:
        menor = c

    # Determinar el mayor
    if a >= b and a >= c:       # Condición Simple
        mayor = a
    elif b >= a and b >= c:     # Selección Múltiple
        mayor = b
    else:
        mayor = c

    return menor, mayor

numero_menor, numero_mayor = encontrar_menor_mayor(num1, num2, num3)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — MENOR Y MAYOR DE TRES NÚMEROS")
print("=" * 50)
print(f"  Número 1   : {num1}")
print(f"  Número 2   : {num2}")
print(f"  Número 3   : {num3}")
print(f"  Menor      : {numero_menor}")
print(f"  Mayor      : {numero_mayor}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 3 — TRIBUTACIÓN DE IMPUESTOS**
Tipo de condición: Condición Compuesta (if / else con and)

Enunciado: Para tributar un determinado impuesto se debe ser mayor de 18 años y tener unos ingresos iguales o superiores a S/3000 mensuales. Escribir un programa que pregunte al usuario su edad y sus ingresos mensuales y muestre por pantalla si el usuario tiene que tributar o no y mostrar el monto a tributar que siempre es el 5% del ingreso mensual.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 3 — TRIBUTACIÓN DE IMPUESTOS          ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Condición Compuesta (if / else con and)     ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
edad             = 25       # int   | edad del usuario en años
ingreso_mensual  = 3500.0   # float | ingreso mensual en soles
tasa_impuesto    = 0.05     # float | tasa de tributación (5%)
edad_minima      = 18       # int   | edad mínima para tributar
ingreso_minimo   = 3000.0   # float | ingreso mínimo para tributar

# --- Cuerpo del algoritmo ---
# Condición Compuesta: if / else con operador and
def calcular_tributacion(edad, ingreso, tasa, edad_min, ingreso_min):
    """Determina si el usuario tributa y calcula el monto."""
    if edad > edad_min and ingreso >= ingreso_min:   # Condición Compuesta
        tributa      = True
        monto_tributo = ingreso * tasa
    else:
        tributa      = False
        monto_tributo = 0.0
    return tributa, monto_tributo

tributa, monto_tributo = calcular_tributacion(
    edad, ingreso_mensual, tasa_impuesto, edad_minima, ingreso_minimo
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — TRIBUTACIÓN DE IMPUESTOS")
print("=" * 50)
print(f"  Edad              : {edad} años")
print(f"  Ingreso mensual   : S/ {ingreso_mensual:.2f}")
if tributa:                                          # Condición Simple
    print(f"  Estado            : DEBE TRIBUTAR")
    print(f"  Monto a tributar  : S/ {monto_tributo:.2f}")
else:
    print(f"  Estado            : NO TRIBUTA")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 4 — COLEGIATURA ESCOLAR CON DESCUENTO**
Tipo de condición: Condición Compuesta (if / else)

Enunciado: En una escuela la colegiatura de los alumnos se determina según el número de materias que cursan (Solicitar al usuario). El costo de todas las materias es el mismo S/100. Se ha establecido un programa para estimular a los alumnos, el cual consiste en lo siguiente: si el promedio obtenido por un alumno en el último periodo es mayor o igual que 18, se le hará un descuento del 50% sobre la mitad de los cursos; si el promedio obtenido es menor que 18 deberá pagar la colegiatura completa. Obtener cuanto debe pagar un alumno.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 4 — COLEGIATURA ESCOLAR               ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Condición Compuesta (if / else)             ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
num_materias    = 6      # int   | número de materias que cursa
promedio        = 19.0   # float | promedio del último período
costo_materia   = 100.0  # float | costo por materia en soles
tasa_descuento  = 0.50   # float | descuento del 50%

# --- Cuerpo del algoritmo ---
# Condición Compuesta: if / else
def calcular_colegiatura(materias, promedio, costo, tasa):
    """Calcula la colegiatura con o sin descuento según promedio."""
    colegiatura_completa = materias * costo
    if promedio >= 18:                          # Condición Compuesta
        mitad_cursos     = materias // 2
        descuento        = (mitad_cursos * costo) * tasa
        colegiatura_pagar = colegiatura_completa - descuento
        tipo             = "CON DESCUENTO (50% en mitad de cursos)"
    else:
        descuento        = 0.0
        colegiatura_pagar = colegiatura_completa
        tipo             = "PAGO COMPLETO"
    return colegiatura_completa, descuento, colegiatura_pagar, tipo

total_sin_desc, monto_desc, total_pagar, tipo_pago = calcular_colegiatura(
    num_materias, promedio, costo_materia, tasa_descuento
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COLEGIATURA ESCOLAR")
print("=" * 50)
print(f"  Materias           : {num_materias}")
print(f"  Promedio           : {promedio:.2f}")
print(f"  Tipo de pago       : {tipo_pago}")
print(f"  Colegiatura base   : S/ {total_sin_desc:.2f}")
print(f"  Descuento          : S/ {monto_desc:.2f}")
print(f"  TOTAL A PAGAR      : S/ {total_pagar:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 5 — TIPO DE TRIÁNGULO**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: Determinar mediante el ingreso de los 3 lados de un triángulo, si este es Equilátero (Mostrar el perímetro del), Isósceles (Indicar que lados son Iguales) o Escaleno (triángulo mostrar el área del triángulo).


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 5 — TIPO DE TRIÁNGULO                 ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

import math  # Para calcular raíz cuadrada (área de Herón)

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
lado_a = 5.0   # float | primer lado del triángulo
lado_b = 5.0   # float | segundo lado del triángulo
lado_c = 8.0   # float | tercer lado del triángulo

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else
def clasificar_triangulo(a, b, c):
    """Clasifica el triángulo y calcula medida correspondiente."""
    if a == b == c:                         # Condición Simple: Equilátero
        tipo      = "EQUILÁTERO"
        perimetro = a + b + c
        detalle   = f"Perímetro = {perimetro:.2f} m"
    elif a == b or b == c or a == c:        # Selección Múltiple: Isósceles
        tipo = "ISÓSCELES"
        if a == b:
            detalle = "Lados iguales: Lado A y Lado B"
        elif b == c:
            detalle = "Lados iguales: Lado B y Lado C"
        else:
            detalle = "Lados iguales: Lado A y Lado C"
    else:                                   # Selección Múltiple: Escaleno
        tipo      = "ESCALENO"
        s         = (a + b + c) / 2        # semiperímetro (Herón)
        area      = math.sqrt(s*(s-a)*(s-b)*(s-c))
        detalle   = f"Área = {area:.2f} m²"
    return tipo, detalle

tipo_triangulo, detalle = clasificar_triangulo(lado_a, lado_b, lado_c)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — TIPO DE TRIÁNGULO")
print("=" * 50)
print(f"  Lado A     : {lado_a:.2f} m")
print(f"  Lado B     : {lado_b:.2f} m")
print(f"  Lado C     : {lado_c:.2f} m")
print(f"  Tipo       : {tipo_triangulo}")
print(f"  Detalle    : {detalle}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 6 — HORA N SEGUNDOS DESPUÉS**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: Elabore un programa que lea la hora (HH MM SS) y muestre por pantalla la hora N segundos después.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 6 — HORA N SEGUNDOS DESPUÉS           ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
horas        = 23    # int | hora actual (HH)
minutos      = 58    # int | minutos actuales (MM)
segundos     = 45    # int | segundos actuales (SS)
n_segundos   = 120   # int | segundos a sumar

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else para manejo de desbordamientos
def sumar_segundos(hh, mm, ss, n):
    """Calcula la hora resultante después de sumar N segundos."""
    total_segundos = hh * 3600 + mm * 60 + ss + n   # total en segundos

    # Extraer nueva hora, minutos y segundos
    nueva_hora     = (total_segundos // 3600) % 24   # Selección Múltiple
    nuevo_minuto   = (total_segundos % 3600) // 60
    nuevo_segundo  = total_segundos % 60

    # Determinar si hubo cambio de día
    if total_segundos >= 86400:                       # Condición Simple
        cambio_dia = "SÍ (pasó medianoche)"
    elif total_segundos >= 3600:                      # Selección Múltiple
        cambio_dia = "NO"
    else:
        cambio_dia = "NO"

    return nueva_hora, nuevo_minuto, nuevo_segundo, cambio_dia

nueva_h, nuevo_m, nuevo_s, cambio = sumar_segundos(
    horas, minutos, segundos, n_segundos
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — HORA N SEGUNDOS DESPUÉS")
print("=" * 50)
print(f"  Hora inicial  : {horas:02d}:{minutos:02d}:{segundos:02d}")
print(f"  Segundos a sumar : {n_segundos} s")
print(f"  Hora final    : {nueva_h:02d}:{nuevo_m:02d}:{nuevo_s:02d}")
print(f"  Cambio de día : {cambio}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 7 — COBRO DE LLAMADA TELEFÓNICA**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: Calcular el pago de una llamada, el cobro es por el tiempo que ésta dura, de tal forma que los primeros cinco minutos cuestan S/ 1.00 c/u, los siguientes tres, S/0.80 c/u, los siguientes dos minutos, S/0.70 c/u, y a partir del décimo minuto, S/0.50 c/u. Además, se carga un impuesto de 3 % cuando es domingo, y 5% si es domingo de turno nocturno. Realice un algoritmo para determinar cuánto debe pagar por cada concepto una persona que realiza una llamada.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 7 — COBRO DE LLAMADA TELEFÓNICA       ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
duracion_min    = 12     # int  | duración de la llamada en minutos
es_domingo      = True   # bool | True si es domingo
es_nocturno     = False  # bool | True si es turno nocturno (domingo)

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else por tramos de minutos
def calcular_llamada(minutos, domingo, nocturno):
    """Calcula el costo de una llamada según duración e impuestos."""
    costo = 0.0

    # Tramo 1: primeros 5 minutos a S/1.00
    if minutos >= 5:                              # Condición Simple
        costo += 5 * 1.00
    else:
        costo += minutos * 1.00

    # Tramo 2: minutos 6 al 8 a S/0.80
    if minutos > 5:                               # Condición Compuesta
        mins_tramo2 = min(minutos - 5, 3)
        costo += mins_tramo2 * 0.80

    # Tramo 3: minutos 9 al 10 a S/0.70
    if minutos > 8:                               # Selección Múltiple
        mins_tramo3 = min(minutos - 8, 2)
        costo += mins_tramo3 * 0.70

    # Tramo 4: a partir del minuto 11 a S/0.50
    if minutos > 10:                              # Selección Múltiple
        mins_tramo4 = minutos - 10
        costo += mins_tramo4 * 0.50

    # Aplicar impuesto según día
    if domingo and nocturno:                      # Condición Compuesta
        impuesto     = costo * 0.05
        tipo_impuesto = "5% (domingo nocturno)"
    elif domingo:                                 # Selección Múltiple
        impuesto     = costo * 0.03
        tipo_impuesto = "3% (domingo)"
    else:
        impuesto     = 0.0
        tipo_impuesto = "Sin impuesto"

    total = costo + impuesto
    return costo, impuesto, tipo_impuesto, total

costo_base, impuesto, tipo_imp, total = calcular_llamada(
    duracion_min, es_domingo, es_nocturno
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COBRO DE LLAMADA TELEFÓNICA")
print("=" * 50)
print(f"  Duración          : {duracion_min} minutos")
print(f"  Costo base        : S/ {costo_base:.2f}")
print(f"  Impuesto          : {tipo_imp} → S/ {impuesto:.2f}")
print(f"  TOTAL A PAGAR     : S/ {total:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 8 — ORDENAR TRES NÚMEROS              ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
num1   = 40      # int | primer número (deben ser diferentes)
num2   = 15      # int | segundo número
num3   = 72      # int | tercer número
orden  = "A"     # str | "A" = ascendente, "D" = descendente

# --- Cuerpo del algoritmo ---
# Validar que los números sean diferentes
def son_diferentes(a, b, c):
    """Verifica que los tres números sean distintos entre sí."""
    return a != b and b != c and a != c

# Ordenar y mostrar según opción elegida
def ordenar_numeros(a, b, c, orden):
    """Ordena tres números en forma ascendente o descendente."""
    lista = sorted([a, b, c])                 # Orden ascendente base

    if orden.upper() == "A":                  # Condición Compuesta
        resultado   = lista
        tipo_orden  = "ASCENDENTE"
    elif orden.upper() == "D":                # Selección Múltiple
        resultado   = lista[::-1]
        tipo_orden  = "DESCENDENTE"
    else:                                     # Selección Múltiple
        resultado   = lista
        tipo_orden  = "OPCIÓN INVÁLIDA — se muestra ascendente"
    return resultado, tipo_orden

# Verificar que los números sean diferentes
if son_diferentes(num1, num2, num3):          # Condición Simple
    numeros_ordenados, tipo = ordenar_numeros(num1, num2, num3, orden)
    valido = True
else:
    valido = False

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — ORDENAR TRES NÚMEROS")
print("=" * 50)
if valido:                                    # Condición Simple
    print(f"  Números ingresados : {num1}, {num2}, {num3}")
    print(f"  Orden solicitado   : {tipo}")
    print(f"  Números ordenados  : {numeros_ordenados[0]}, {numeros_ordenados[1]}, {numeros_ordenados[2]}")
else:
    print("  ERROR: Los números deben ser diferentes entre sí.")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 9 — COMPRA DE CAMISAS CON DESCUENTO**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: Permita calcular el total a pagar por la compra de N camisas. Si se compran entre 1 a 4 camisas se aplica un descuento del 12.5%, si se compra una cantidad comprendida entre 5 y 8 camisas se aplica un descuento del 20% y si se compran cantidades mayores, se aplica un descuento del 31.5% sobre el total de la compra. Debe imprimirse en pantalla la compra final sin descuento, monto del descuento y la compra con descuento.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 9 — COMPRA DE CAMISAS CON DESCUENTO  ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
num_camisas     = 6       # int   | cantidad de camisas a comprar
precio_camisa   = 80.0    # float | precio unitario por camisa en soles

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else por tramos de cantidad
def calcular_compra_camisas(cantidad, precio):
    """Calcula el total con descuento según cantidad de camisas."""
    total_sin_desc = cantidad * precio

    if 1 <= cantidad <= 4:                    # Condición Simple
        tasa_descuento = 0.125
        rango          = "1 a 4 camisas"
    elif 5 <= cantidad <= 8:                  # Selección Múltiple
        tasa_descuento = 0.20
        rango          = "5 a 8 camisas"
    else:                                     # Selección Múltiple
        tasa_descuento = 0.315
        rango          = "más de 8 camisas"

    monto_descuento = total_sin_desc * tasa_descuento
    total_con_desc  = total_sin_desc - monto_descuento
    return total_sin_desc, tasa_descuento, monto_descuento, total_con_desc, rango

sin_desc, tasa, descuento, con_desc, rango = calcular_compra_camisas(
    num_camisas, precio_camisa
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — COMPRA DE CAMISAS")
print("=" * 50)
print(f"  Cantidad            : {num_camisas} camisas ({rango})")
print(f"  Precio unitario     : S/ {precio_camisa:.2f}")
print(f"  Total sin descuento : S/ {sin_desc:.2f}")
print(f"  Descuento ({tasa*100:.1f}%)    : S/ {descuento:.2f}")
print(f"  TOTAL A PAGAR       : S/ {con_desc:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```

## **EJERCICIO N° 10 — LLAMADAS INTERNACIONALES POR ZONA**
Tipo de condición: Selección Múltiple (if / elif / else)

Enunciado: El costo de llamadas telefónicas internacionales depende de clave (identifica a la zona geográfica) en la que se encuentra el destino y del número de minutos hablados. En la siguiente tabla se presenta el costo del minuto por zona: Código 10 América del norte S/2.2, Código 12 América del centro S/2.5, Código 20 América del sur S/1.2, Código 22 Asia S/3.5, Código 30 Europa S/3.0, Código 32 África S/3.2. Construya un diagrama de flujo que permita mostrar el nombre del área y calcular e imprimir el costo total de una llamada y el nombre de la zona a la cual llamó.


```python
# ╔══════════════════════════════════════════════════════╗
# ║  EJERCICIO N° 10 — LLAMADAS INTERNACIONALES         ║
# ║  Curso : Fundamentos de Programación IDL2            ║
# ║  Tema  : Estructuras Condicionales                   ║
# ║  Tipo  : Selección Múltiple (if / elif / else)       ║
# ╚══════════════════════════════════════════════════════╝

# ═══════════ INICIO ═══════════

# --- Declaración e inicialización de variables ---
codigo_zona     = 20    # int   | código de zona geográfica
minutos_llamada = 8     # int   | minutos de llamada

# --- Tabla de zonas (referencia) ---
# CODIGO | ZONA              | COSTO x MIN (S/)
#   10   | América del Norte |   2.2
#   12   | América del Centro|   2.5
#   20   | América del Sur   |   1.2
#   22   | Asia              |   3.5
#   30   | Europa            |   3.0
#   32   | África            |   3.2

# --- Cuerpo del algoritmo ---
# Selección Múltiple: if / elif / else por código de zona
def calcular_llamada_internacional(codigo, minutos):
    """Calcula el costo de llamada internacional según zona y minutos."""
    if codigo == 10:                          # Condición Simple
        zona   = "América del Norte"
        costo_min = 2.2
    elif codigo == 12:                        # Selección Múltiple
        zona   = "América del Centro"
        costo_min = 2.5
    elif codigo == 20:                        # Selección Múltiple
        zona   = "América del Sur"
        costo_min = 1.2
    elif codigo == 22:                        # Selección Múltiple
        zona   = "Asia"
        costo_min = 3.5
    elif codigo == 30:                        # Selección Múltiple
        zona   = "Europa"
        costo_min = 3.0
    elif codigo == 32:                        # Selección Múltiple
        zona   = "África"
        costo_min = 3.2
    else:                                     # Selección Múltiple
        zona   = "ZONA NO RECONOCIDA"
        costo_min = 0.0

    costo_total = minutos * costo_min
    return zona, costo_min, costo_total

nombre_zona, costo_por_min, costo_total = calcular_llamada_internacional(
    codigo_zona, minutos_llamada
)

# --- Mostrar resultados ---
print("=" * 50)
print("   RESULTADO — LLAMADA INTERNACIONAL")
print("=" * 50)
print(f"  Código de zona    : {codigo_zona}")
print(f"  Zona geográfica   : {nombre_zona}")
print(f"  Minutos hablados  : {minutos_llamada} min")
print(f"  Costo por minuto  : S/ {costo_por_min:.2f}")
print(f"  COSTO TOTAL       : S/ {costo_total:.2f}")
print("=" * 50)

# ═══════════ FIN ═══════════
```
